```python
import numpy as np
import pandas as pd

# 1. Create a simple dataset
data = {
    "Name": ["Aarav", "Priya", "Rohan", "Sneha", "Priya"],
    "Marks": [85, np.nan, 90, 75, np.nan],
    "City": ["Pune", "Mumbai", "Pune", np.nan, "Mumbai"],
}

df = pd.DataFrame(data)
print("Original Data:")
print(df)

# 2. Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# 3. Handle missing values
# Fill missing Marks with mean
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
# Fill missing City with 'Unknown'
df["City"] = df["City"].fillna("Unknown")

# 4. Remove duplicate rows
df = df.drop_duplicates()

print("\nData after cleaning:")
print(df)

# 5. Filter data using .loc[]
# Students with marks greater than 80
top_students = df.loc[df["Marks"] > 80, ["Name", "Marks"]]
print("\nStudents with marks > 80:")
print(top_students)

# Students from Pune
pune_students = df.loc[df["City"] == "Pune"]
print("\nStudents from Pune:")
print(pune_students)
```
