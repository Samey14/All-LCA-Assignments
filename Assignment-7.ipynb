```python
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

csv_filename = "sample_data_assignment_07.csv"

# Step 0: Ensure the CSV file exists (creates it automatically if missing)
if not os.path.exists(csv_filename):
  csv_content = """Student,Maths,Python,Attendance,City
Amit,75,80,85,Pune
Bhavna,82,78,90,Mumbai
Chetan,68,70,78,Pune
Diya,90,92,95,Nashik
Esha,72,75,80,Mumbai
Farhan,85,88,92,Pune"""
  with open(csv_filename, "w") as file:
    file.write(csv_content)

# Step 1: Load the CSV file
df = pd.read_csv(csv_filename)

# Step 2: Display the dataset & average metrics
print("Dataset:")
print(df)
print("-" * 50)
print(f"Average Maths Marks     : {df['Maths'].mean():.2f}")
print(f"Average Python Marks    : {df['Python'].mean():.2f}")
print(f"Average Attendance Rate : {df['Attendance'].mean():.2f}%")
print("-" * 50)

# Step 3: Line Plot (Python Marks)
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o", color="blue", linewidth=2)
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.tight_layout()
plt.show()

# Step 4: Bar Plot (Maths Marks)
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Maths"], color="teal", edgecolor="black")
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.tight_layout()
plt.show()

# Step 5: Scatter Plot (Maths vs Python Marks)
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Maths"], df["Python"], color="crimson", s=70)
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.tight_layout()
plt.show()

# Step 6: Correlation Matrix & Heatmap
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("Correlation Matrix:")
print(correlation)
print("-" * 50)

plt.figure(figsize=(6.5, 5))
sns.heatmap(
    correlation, annot=True, cmap="Blues", vmin=-1, vmax=1, fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

# ==========================================
# Challenge / Extension Tasks
# ==========================================

# Step 7: Histogram (Python Marks Distribution)
plt.figure(figsize=(7, 4.5))
plt.hist(
    df["Python"],
    bins=5,
    color="skyblue",
    edgecolor="black",
    label="Python Scores",
)
plt.title("Distribution of Python Marks")
plt.xlabel("Marks Range")
plt.ylabel("Number of Students")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

# Step 8: Box Plot (Maths & Python Marks)
plt.figure(figsize=(7, 4.5))
sns.boxplot(data=df[["Maths", "Python"]], palette="Set2")
plt.title("Distribution Comparison: Maths vs Python")
plt.ylabel("Marks")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

# Step 9: Pie Chart (City-wise Student Distribution)
city_counts = df["City"].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(
    city_counts,
    labels=city_counts.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=["#5DADE2", "#58D68D", "#F5B041"],
)
plt.title("City-wise Student Distribution")
plt.tight_layout()
plt.show()
```
