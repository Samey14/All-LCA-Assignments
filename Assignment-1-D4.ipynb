```python
#Create list:

student_list=["Samey","Vedant","Pushpak","Aditya"]
print("\nStudent List:",student_list)

#Create tuple:

student_tuple=("Samey","Vedant","Pushpak","Aditya")
print("\nStudent Tuple:",student_tuple)

#Create dictionary:

student_dict={50:"Samey",51:"Vedant",52:"Pushpak",53:"Aditya"}
print("\nStudent Dictionary:",student_dict)
```

```python
#Adding element in dictionary:
student_dict[54]="Anish"
print("\nAfter Adding:",student_dict)

#Updating element in dictionary:
student_dict[50]="Avaneesh"
print("\nAfter Updating:",student_dict)

#Deleting element fron dictionary:
del student_dict[51]
print("\nAfter Deleting:",student_dict)
```
